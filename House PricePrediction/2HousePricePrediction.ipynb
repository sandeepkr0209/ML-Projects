#Splitting the data

```python
X = df.drop('price', axis=1)
y = df['price']
```

# Splitting the data further for training

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=5)
```

#Import models


```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
dt = DecisionTreeRegressor()
rg = RandomForestRegressor(n_estimators=500)
```

# Performing Encoding on categorical and numerical data with different method

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

numeric_features = [0,1,2,3,9]
ohe_features = [4,5,6,7,8,10]
ordinal_features = [11]

preprocess = ColumnTransformer(transformers = [
    ('num', StandardScaler(), numeric_features),
    ('ohe', OneHotEncoder(drop='first'), ohe_features),
    ('ord', OrdinalEncoder(categories=[['furnished','semi-furnished','unfurnished']]), ordinal_features),
])
```

# Creating Pipeline

```python
pipe1 = Pipeline(steps=[
    ('preprocess', preprocess),
    ('model', dt)
])

pipe2 = Pipeline(steps=[
    ('preprocess', preprocess),
    ('model', rg)
])
```

#Training the model

```python
pipe1.fit(X_train, y_train)
```

```python
pipe2.fit(X_train, y_train)
```

```python
y_pred1 = pipe1.predict(X_test)
y_pred2 = pipe2.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

#Evaluating the model performance

```python
print('Decision Tree Evaluation :')
print('R2 Score: ', r2_score(y_test, y_pred1))
print('MSE: ', mean_squared_error(y_test, y_pred1))
print('MAE :', mean_absolute_error(y_test, y_pred1))
```

```python
print('Random Forest Evaluation :')
print('R2 Score: ', r2_score(y_test, y_pred2))
print('MSE: ', mean_squared_error(y_test, y_pred2))
print('MAE :', mean_absolute_error(y_test, y_pred2))
```

```python
from sklearn.linear_model import LinearRegression
```

#Baseline Model testing

```python
lr = LinearRegression()

pipe3 = Pipeline(steps=[
    ('preprocess', preprocess),
    ('model', lr)
])

pipe3.fit(X_train, y_train)
```

```python
y_pred3 = pipe3.predict(X_test)
```

```python
print('Linear Regression Evaluation :')
print('R2 Score: ', r2_score(y_test, y_pred2))
print('MSE: ', mean_squared_error(y_test, y_pred2))
print('MAE :', mean_absolute_error(y_test, y_pred2))
```
